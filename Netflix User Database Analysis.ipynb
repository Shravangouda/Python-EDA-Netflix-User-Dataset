## 📈 Important Charts

###  Subscription Type

```python
df.head()
```

```python
counts=df['Subscription Type'].value_counts()
plt.figure(figsize=(7,5))

plt.bar(counts.index,counts.values, color='Red')
ax=plt.gca()
for container in ax.containers:
    ax.bar_label(container)
plt.xlabel('Subscription Type', color='Black')
plt.ylabel('The no.of Subscriptions',color='Black')
plt.tick_params(axis='both', colors='Red')
plt.tight_layout()
plt.show()
```

### Revenue by Subscription Type

```python
revenue=df.groupby('Subscription Type')['Monthly Revenue'].sum()
revenue
```

```python
plt.figure(figsize=(8,5))
plt.barh(revenue.index, revenue.values, color='Black')
plt.xlabel('Monthly Revenue',color='Red')
plt.ylabel('Subscription Type',color='Red')
plt.tick_params(axis='both', colors='red')
ax=plt.gca()
for container in ax.containers:
    ax.bar_label(container)
plt.tight_layout()
plt.show()
```

### Gender Distribution

```python
Gender=df['Gender'].value_counts()
```

```python
Gender
```

```python
import matplotlib.pyplot as plt

# Create pie chart
plt.figure(figsize=(6, 6))
plt.pie(Gender, labels=Gender.index, autopct='%1.1f%%', startangle=240, explode=[0.02, 0])
plt.title("Gender Distribution")
plt.show()
```

### Monthly Revenue Distribution

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
# Plot actual Monthly Revenue values
sns.histplot(df['Monthly Revenue'], bins=5, color='Red')

plt.title('Monthly Revenue Distribution')
plt.xlabel('Monthly Revenue ($)')
plt.ylabel('User Count')
plt.show()
```

### Revenue by Country

```python
con_ren=df.groupby('Country')['Monthly Revenue'].sum()
con_ren
```

```python
plt.figure(figsize=(20,10))
sns.barplot(x=con_ren.index,y=con_ren.values,color='Red')
plt.xlabel('Country',fontsize=20)
plt.ylabel('Total Revenue ($)',fontsize=20)
plt.show()
```

### Device Usage

```python
device_conut=df['Device'].value_counts()
device_conut
```

```python
plt.figure(figsize=(10,2))
plt.barh(y=device_conut.index,width=device_conut.values,color='Red')
ax=plt.gca()
for container in ax.containers:
    ax.bar_label(container)

# Add 15% padding to the right side
ax.set_xlim(right=device_conut.values.max() * 1.15) 
plt.tight_layout()
plt.show()
```

## Country-wise Subscription Type

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Country', hue='Subscription Type', palette='Set1')

plt.title('Subscription Type Distribution by Country')
plt.xlabel('Country')
plt.ylabel('User Count')
plt.xticks(rotation=45)
plt.legend(title='Subscription Type')
plt.show()
```

##  Gender vs Subscription Type(Count Plot (Hue))

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Gender', hue='Subscription Type', palette='Set1')

plt.title('Subscription Type Distribution by Country')
plt.xlabel('Gender')
plt.ylabel('User Count')
plt.xticks(rotation=45)
plt.legend(title='Subscription Type')
plt.show()
```
